### feature engineering

- we are going to create col such as brand and brand model
- ram and storage from storage col
- primary_rear_camera and no of rear camera from back_camera
- primary_front_camera from fron_camera
- battery_capacity and charging_type from battery col
- display_size and display_type form display col
- exact anTuTu score

```python
df.sample(1)
```

```python
def len_of_list(l):
    return len(l)
```

```python
brand = df['brand_name'].str.split(' ').str[0]
df.insert(0, 'brand', value=brand)
```

```python
ram = df['storage'].str.split('|').str[0].str.split(' ').str[0]
df.insert(4, 'ram', value=ram)
```

```python
storage_capacity = df['storage'].str.split('|').str[1].str.split(' ').str[1]
df.insert(5, 'storage_capacity', value=storage_capacity)
```

```python
primary_rear_camera = df['back_camera'].str.split('+').str[0].str.split(' ').str[0]
df.insert(7, 'primary_rear_camera', value=primary_rear_camera)
```

```python
front_camera = df['fron_camera'].str.split(' ').str[0]
df.insert(10, 'front_camera', value=front_camera)
```

```python
battery_capacity  = df['battery'].str.split('|').str[0].str.split(' ').str[0]
df.insert(12, 'battery_capacity', value=battery_capacity)
```

```python
charging_type = df['battery'].str.split('|').str[1]
df.insert(13, 'charging_type', value=charging_type)
```

```python
display_type = df['display'].str.split('|').str[1]
df.insert(15, 'display_type', value=display_type)
```

```python
antutu_score = df['anTuTu_score'].str.split(' ').str[2]
df.insert(17, 'antutu_score', value=antutu_score)
```

```python
df.sample(1)
```

```python
camera_count = df['back_camera'].str.split('+').apply(len_of_list)
df.insert(8, 'camera_count', value=camera_count)
```

```python
list_old_col = ['storage', 'fron_camera', 'battery', 'price', 'anTuTu_score', 'display']
```

```python
df = df.drop(columns=list_old_col)
```

```python
df.sample(5)
```

```python
df.info()
```

```python
df.rename(columns={'brand_name':'model'}, inplace=True)
```

```python
df['ram'] = df['ram'].str.replace('GB','')
df['ram'] = df['ram'].astype(float)
```

```python
df['primary_rear_camera'] = df['primary_rear_camera'].str.replace("MP","")
df['primary_rear_camera'] = df['primary_rear_camera'].astype(float)

df['front_camera'] = df['front_camera'].str.replace("MP","")
df['front_camera'] = df['front_camera'].astype(float)
```

```python
df['battery_capacity'] = df['battery_capacity'].str.replace("mAh","")
df['battery_capacity'] = df['battery_capacity'].astype(int)
```

```python
df['antutu_score'] = df['antutu_score'].str.replace(',', '', regex=True)
df['antutu_score'] = df['antutu_score'].astype('Int64')
```

```python
df.to_csv("cleaned_df.csv", index=False)
```
