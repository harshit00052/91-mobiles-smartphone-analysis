### feature engineering

- we are going to create col such as brand and brand model
- ram and storage from storage col
- primary_rear_camera and no of rear camera from back_camera
- primary_front_camera from fron_camera
- battery_capacity and charging_type from battery col
- display_size and display_type form display col
- exact anTuTu score

```python
def len_of_list(l):
    return len(l)
```

```python
df['brand'] = df['brand_name'].str.split(' ').str[0]
df['ram'] = df['storage'].str.split('|').str[0].str.split(' ').str[0]
df['storage_capacity'] = df['storage'].str.split('|').str[1].str.split(' ').str[1]
df['primary_rear_camera'] = df['back_camera'].str.split('+').str[0].str.split(' ').str[0]
df['front_camera'] = df['fron_camera'].str.split(' ').str[0]
df['battery_capacity']  = df['battery'].str.split('|').str[0].str.split(' ').str[0]
df['charging_type'] = df['battery'].str.split('|').str[1]
df['display_type'] = df['display'].str.split('|').str[1]
df['antutu_score'] = df['anTuTu_score'].str.split(' ').str[2]
```

```python

df['camera_count'] = df['back_camera'].str.split('+').apply(len_of_list)
```

```python
list_old_col = ['storage', 'back_camera', 'fron_camera', 'battery', 'display', 'anTuTu_score']
```

```python
df = df.drop(columns=list_old_col)
```
